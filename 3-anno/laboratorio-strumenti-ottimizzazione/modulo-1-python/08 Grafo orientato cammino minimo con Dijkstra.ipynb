# Grafo orientato risolto con Algoritmo di Dijkstra
Leggiamo i dati di un grafo orientato pesato da file e li trasformiamo in una matrice di adiacenza.
Poi applichiamo l'algoritmo di Dijkstra per trovare il cammino minimo da una sorgente $s$ a un terminale $t$.

## Cosa facciamo?
1. Apriamo un file locale e leggiamo numero di vertici $n$ e numero di archi $m$.
2. Memorizziamo archi e costi in due liste.
3. Costruiamo la matrice di adiacenza.
4. Eseguiamo Dijkstra da $s$ a $t$.
5. Ricostruiamo il cammino minimo tramite i predecessori.

Il grafo e orientato, quindi l'arco $i \to j$ non implica l'arco $j \to i$.

```mermaid
graph LR
    n1(("1")) -->|1| n2(("2"))
    n2 -->|2| n4(("4"))
    n3(("3")) -->|10| n4
    n3 -->|1| n5(("5"))
    n4 -->|3| n5
    n1 -->|4| n3
```

Nel file dati usato in questo esercizio avremo:

```
5 6
1 2 1
2 4 2
3 4 10
3 5 1
4 5 3
1 3 4
```

La matrice di adiacenza $A$ con infinito dove non esiste arco e:

$$
A =
\begin{pmatrix}
0 & 1 & 4 & \infty & \infty \\
\infty & 0 & \infty & 2 & \infty \\
\infty & \infty & 0 & 10 & 1 \\
\infty & \infty & \infty & 0 & 3 \\
\infty & \infty & \infty & \infty & 0
\end{pmatrix}
$$

````python
import numpy as np

input_file = open("08 Grafo orientato.txt", "r")
riga = input_file.readline().split()

n = int(riga[0])
m = int(riga[1])

print("n=", n)
print("m=", m)

archi_letti = []
costi = []

for _ in range(m):
    read_string = input_file.readline()
    print("\n", read_string)

    lista = read_string.split()
    i = int(lista[0])
    j = int(lista[1])
    c = int(lista[2])

    archi_letti.append([i, j])
    costi.append(c)
    print("letto arco", i, "-", j)

input_file.close()

print(archi_letti)
print(costi)

infinito = 999999
adj = np.full((n, n), infinito, dtype=int)

for i in range(n):
    adj[i][i] = 0

for a in range(m):
    i = archi_letti[a][0]
    j = archi_letti[a][1]
    adj[i - 1][j - 1] = costi[a]

print(adj)
````

L'algoritmo di Dijkstra risolve il problema del cammino minimo da una sorgente $s$ a tutti gli altri vertici, in particolare verso un terminale $t$.

### Come funziona l'algoritmo
L'algoritmo segue una logica greedy e assegna etichette permanenti in modo progressivo:
1. Partenza: si inizializzano distanza e predecessore di ogni vertice.
2. Selezione: si sceglie tra i vertici non raggiunti quello con distanza minima.
3. Aggiornamento: si provano i rilassamenti sui vertici adiacenti.
4. Termine: si conclude quando $t$ e raggiunto o non esistono vertici raggiungibili.

---

### Linea guida per implementare l'esercizio
1. Definisci una classe Vertex con: indice, nome, predecessore, stato reached, distanza l.
2. Crea la lista dei vertici V.
3. Inizializza il vertice sorgente con distanza 0.
4. Ripeti selezione del minimo e aggiornamento dei vicini.
5. Ricostruisci il cammino minimo tramite i predecessori.

````python
class Vertex:
    def __init__(self):
        self.index = None
        self.name = None
        self.pred = None
        self.reached = False  # True se il vertice e in S
        self.l = None


V = []
for i in range(n):
    v = Vertex()
    v.name = "v_" + str(i + 1)
    v.index = i
    v.pred = None
    v.reached = False
    v.l = infinito
    V.append(v)

source_index = 0
target_index = n - 1

vertex_s = V[source_index]
vertex_t = V[target_index]

vertex_s.reached = True
vertex_s.l = 0
vertex_s.pred = vertex_s.index

for v in V:
    if v.index != vertex_s.index:
        v.l = adj[vertex_s.index][v.index]
        v.pred = vertex_s.index

for v in V:
    print(v.name, "pred:", v.pred, "reached:", v.reached, "L:", v.l)
````

````python
while vertex_t.reached == False:
    minimo = infinito
    imin = None

    for i, v in enumerate(V):
        if (v.reached == False) and (v.l < minimo):
            minimo = v.l
            imin = i

    if imin is None or minimo >= infinito:
        print("Non esiste cammino tra sorgente e terminale!")
        break

    print("prossimo vertice:", imin + 1)

    vertice = V[imin]
    vertice.reached = True

    for w in V:
        if (w.reached == False) and (adj[vertice.index][w.index] < infinito):
            nuovo = vertice.l + adj[vertice.index][w.index]
            if nuovo < w.l:
                w.pred = vertice.index
                w.l = nuovo
````

````python
for v in V:
    print("vertex:", v.name, "pred:", v.pred, "length:", v.l)

if vertex_t.l is None or vertex_t.l >= infinito:
    print("Cammino minimo non trovato")
else:
    path = [vertex_t.name]
    i = vertex_t.index

    while i != vertex_s.index:
        predecessore = V[i].pred
        path.append(V[predecessore].name)
        i = predecessore

    path.reverse()
    print("Cammino minimo:", path)
    print("Costo totale:", vertex_t.l)
````
