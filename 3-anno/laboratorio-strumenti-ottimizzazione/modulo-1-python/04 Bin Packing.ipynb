# Bin Packing


```python
import pyomo.environ as pyo

# dati
items = ['item1', 'item2', 'item3','item4','item5']
bins = ['bin1', 'bin2', 'bin3','bin4','bin5']
p = [7,4,1,4,4]
w_max = 10

model = pyo.ConcreteModel("Bin Packing Problem")

# variabili
model.y = pyo.Var(bins, within=pyo.Binary)
model.x = pyo.Var(bins, items, within=pyo.Binary)

pesi = dict(zip(items,p))

# FUNZIONE OBIETTIVO
def funzione_obiettivo(m):
    return sum(m.y[i] for i in bins)
model.obiettivo = pyo.Objective(rule=funzione_obiettivo, sense=pyo.minimize)

# VINCOLO 1: Ogni oggetto in esattamente 1 bin, non duplicati
def vincolo_tutti_oggetti(m, j):
    return sum(m.x[i,j] for i in bins) == 1
model.vincolo1 = pyo.Constraint(items, rule=vincolo_tutti_oggetti)

# VINCOLO 2: La somma dei pesi nel bin i non deve superare la capacità
def vincolo_pesi(m, i):
    return sum(pesi[j] * m.x[i,j] for j in items) <= w_max*m.y[i]

# CORRETTO: Nome cambiato in vincolo2, iteratore cambiato in bins, rule cambiata in vincolo_pesi
model.vincolo2 = pyo.Constraint(bins, rule=vincolo_pesi)

# SOLUZIONE
ottimizzatore = pyo.SolverFactory('glpk')
risultato = ottimizzatore.solve(model, tee=False)

model.display()
```

# Bin Packing con dati Random
Ora proviamo a fare un modello che non abbia più solo dati hardcoded ma generati randomicamente a ogni istanza.
- 10 Bins
- Capacità di ogni Bin di 10
- 10 Oggetti
- Pesi random degli oggetti

```python
import pyomo.environ as pyo
import random as rnd

# dati
#10 bins e 10 items
n=10
items = []
bins = []
p = []
rnd.random()
w_max = 10

for i in range(1, n+1):
    items.append('item'+str(i))
    bins.append('bin'+str(i))
    p.append(1+int(w_max*rnd.random()))
print(items)
print(p)
print(bins)
model = pyo.ConcreteModel("Bin Packing Problem")

# variabili
model.y = pyo.Var(bins, within=pyo.Binary)
model.x = pyo.Var(bins, items, within=pyo.Binary)

pesi = dict(zip(items,p))

# FUNZIONE OBIETTIVO
def funzione_obiettivo(m):
    return sum(m.y[i] for i in bins)
model.obiettivo = pyo.Objective(rule=funzione_obiettivo, sense=pyo.minimize)

# VINCOLO 1: Ogni oggetto in esattamente 1 bin, non duplicati
def vincolo_tutti_oggetti(m, j):
    return sum(m.x[i,j] for i in bins) == 1
model.vincolo1 = pyo.Constraint(items, rule=vincolo_tutti_oggetti)

# VINCOLO 2: La somma dei pesi nel bin i non deve superare la capacità
def vincolo_pesi(m, i):
    return sum(pesi[j] * m.x[i,j] for j in items) <= w_max*m.y[i]

# CORRETTO: Nome cambiato in vincolo2, iteratore cambiato in bins, rule cambiata in vincolo_pesi
model.vincolo2 = pyo.Constraint(bins, rule=vincolo_pesi)

# SOLUZIONE
ottimizzatore = pyo.SolverFactory('glpk')
risultato = ottimizzatore.solve(model, tee=False)

model.display()
```
